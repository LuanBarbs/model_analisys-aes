```python
import pandas as pd

df = pd.read_csv("../dataset/output/avalia_saida.csv")

modelos = ["LLAMA", "GEMMA", "GPT4"]
mecanismos = ["ER_label", "IP_label", "EX_label"]
classes = [0, 1, 2]

# tabela: uma linha por modelo, contagens 0/1/2 para cada mecanismo
linhas = []
for m in modelos:
    sub = df[df["id"] == m]
    linha = {"Model": m}
    for mec in mecanismos:
        # reindex garante que classes ausentes apareçam como 0
        contagem = sub[mec].value_counts().reindex(classes, fill_value=0)
        for c in classes:
            linha[f"{mec[:2]}_{c}"] = int(contagem[c])
    linhas.append(linha)

tabela = pd.DataFrame(linhas).set_index("Model")
print(tabela)
```
